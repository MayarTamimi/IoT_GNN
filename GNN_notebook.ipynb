```python
import pandas as pd
import requests
from io import StringIO

print("Libraries ready")
```

**• Load the dataset**

```python
url = "https://mcfp.felk.cvut.cz/publicDatasets/IoT-23-Dataset-v2/CTU-IoT-Malware-Capture-34-1/2018-12-21-15-50-14-192.168.1.195-zeek-conn-log.labeled"

response = requests.get(url, timeout=60)

print("Status:", response.status_code)
print("File size:", round(len(response.content) / (1024 * 1024), 2), "MB")
```

```python
text = response.text
lines = [
    line for line in text.splitlines()
    if line.strip() and not line.startswith("#")
]

columns_name = [
    line for line in text.splitlines()
    if line.strip() and line.startswith("#fields")
][0]

columns_name = columns_name.replace("#fields", "").split()

df = pd.read_csv(
    StringIO("\n".join(lines)),
    sep=r"\s+",
    header=None,
    names=columns_name,
    engine="python"
)

print("Rows:", len(df))
print("Columns:", len(df.columns))

```

# **Dataset Summary**

The dataset contains 23,145 network connection records and 23 features. It uses about 4.1 MB of memory.

There are no missing values and no duplicate records.
The dataset is highly imbalanced:

- Malicious: 21,222 records (91.69%)
- Benign: 1,923 records (8.31%)

Most columns are categorical (object), while some network statistics such as ports, packet counts, and byte counts are numeric.

What the dataset represents:

**The dataset represents** IoT network traffic connections. Each row describes one communication session between a source device and a destination device, including information such as:

- source and destination IP addresses
- ports
- protocol
- service
- connection duration
- bytes and packets transferred
- connection state
- traffic label

The label column identifies whether the traffic is Benign or Malicious, while detailed-label provides a more specific traffic or attack category.
